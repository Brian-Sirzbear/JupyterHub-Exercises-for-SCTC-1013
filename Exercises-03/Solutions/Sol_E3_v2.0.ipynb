### Section 2.1: Extracting Information
We can generally separate the process into two parts.
* First, we modify the table such that the information we want is easy to obtain
* Then, we use specialized methods and functions to get that information.

We'll focus on the second step in this Section, and then go to the first step in the following Section.

These 'specialized methods and functions' consist most commonly of `.column`, `.num_rows`, `.take`, and statistical functions like `np.average`, `np.ptp`, and `np.std`. These are summarized in the table below.

|Code|Purpose|Type of Question it is answering|
|--|--|--|
|**`table`**`.column(`**`label`**`)` OR **`table`**`[`**`label`**`]`|Goes into **`table`** and extracts the specified table column as an array.|Anything regarding a statistical function (mean, range, std, etc). You will **almost always** need to use this prior to the statistical function, as they only accept arrays.|
|**`table`**`.num_rows`|Returns the number of rows in **`table`**.|Anything asking for/involving **how many** of a certain thing there is.<br>Depends on the content of the table you're working with.*|
|**`table`**`.take(np.arange(`**`n`**`)`|Returns just the first **`n`** rows of the **`table`**.|Anything asking for some rows of the table, almost always the most extreme rows (those with the highest/lowest values in a column).|
|`np.average(`**`array`**`)` OR `np.mean(`**`array`**`)`|Gets the mean of **`array`**.|Anything asking for/involving the mean.|
|`np.ptp(`**`array`**`)`|Gets the range of **`array`**.|Anything asking for/involving the range (highest value minus lowest value).|
|`np.std(`**`array`**`)`|Gets the standard deviation of **`array`**.|Anything asking for/involving the standard deviation.|
|`sum(`**`array`**`)`|Gets the sum of **`array`**.|Anything asking for/involving the total.|

```python
imdb.pivot('Rating', 'Decade', 'Votes', np.average)
```

As with `.group`, the function must accept only a list/array.

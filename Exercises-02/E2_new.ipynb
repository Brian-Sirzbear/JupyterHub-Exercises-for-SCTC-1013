There's another function used in statistics that is very common, and that is **standard deviation** ($\sigma$). This describes how closely the values are clustered around the average ($\mu$).

$$\sigma = \sqrt{\frac{\Sigma(x - \mu)^2}{n}}$$

|Step|Equivalent Expression|
|--|--|
|The $x$ represents all the terms in the array. So, take your array and subtract the mean from all the terms inside.|$x-\mu$|
|You'll get an array of differences. Square each difference.|$(x-\mu)^2$|
|Add these values up. (The sum is represented by $\Sigma$)|$\Sigma(x-\mu)^2$|
|Now you have a single value. Divide that by the number of terms in the array.|$\frac{\Sigma(x-\mu)^2}{n}$|
|And finally, take the square root.|$\sigma = \sqrt{\frac{\Sigma(x - \mu)^2}{n}}$|
